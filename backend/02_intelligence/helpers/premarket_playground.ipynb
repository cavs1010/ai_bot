## `get_premarket_gap(ticker)`

**Expected output:** dict with `gap_pct`, `prior_close`, `premarket_price`, `direction`  
**Gate 1 block rules:** gap down ≥ 2% → block | gap up ≥ 4% → block  
**Note:** returns `None` if called outside the pre-market window (4:00–9:29 AM ET)

```python
result = get_premarket_gap('NVDA')
if result:
    print(f'prior close:      {result["prior_close"]}')
    print(f'pre-market price: {result["premarket_price"]}')
    print(f'gap:              {result["gap_pct"]:+.2%} ({result["direction"]})')
else:
    print('None returned — no pre-market bars (market may be open or it is a weekend)')
```
