# `helpers/filings.py` — Playground

Manual exploration of `get_recent_8k_filings(ticker, days_back)`.

Fetches recent SEC EDGAR 8-K filings via the public atom feed.  
No API key required. Used by Gate 1 to block tickers with filings today.

```python
# Cell 1 — Setup & import
import sys
sys.path.insert(0, 'backend/02_intelligence/helpers')

from filings import get_recent_8k_filings
```

```python
# Cell 2 — Happy path: 30-day window (guarantees results)
filings = get_recent_8k_filings('AAPL', days_back=30)
print(f'Found {len(filings)} filing(s) in the past 30 days:')
for f in filings:
    print(f"  {f['filed_at']}  {f['form_type']}  {f['title']}")
    print(f"           {f['url']}")
```

```python
# Cell 3 — Parameter variations
# days_back=1: today and yesterday (Gate 1 default)
one_day  = get_recent_8k_filings('AAPL', days_back=1)
# days_back=7: past week
one_week = get_recent_8k_filings('AAPL', days_back=7)
# Different ticker
nvda     = get_recent_8k_filings('NVDA', days_back=30)

print(f'AAPL — past 1 day:  {len(one_day)} filing(s)')
print(f'AAPL — past 7 days: {len(one_week)} filing(s)')
print(f'NVDA — past 30 days: {len(nvda)} filing(s)')
```

```python
# Cell 4 — Failure path
# Unknown ticker: SEC returns 200 with 0 entries → empty list (not None)
bad = get_recent_8k_filings('ZZZZINVALID', days_back=1)
print(f'Unknown ticker result: {bad!r}')  # expect []
assert bad == [], 'Expected empty list'
print('✅ unknown ticker correctly returned []')
```
