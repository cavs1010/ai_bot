Tier 2 nightly scan: reads `data/watchlist.csv` → returns top momentum candidates.

```python
# Cell 1 — Setup & import
import sys
sys.path.insert(0, 'backend/01_scanner')

from momentum_scanner import load_watchlist, calculate_momentum_score, run_scan
import pandas as pd

print('Imports OK')
```

```python
# Cell 2 — Inspect raw watchlist
df = load_watchlist()
if df is None:
    raise FileNotFoundError(f'watchlist not found — run universe_filter.py first')
display(df.head(10))
print(f'\nColumns: {df.columns.tolist()}')
```

```python
# Cell 3 — Run momentum scan and view top candidates
results = run_scan( min_score=3)
display(results)
```

```python
# Cell 4 — Parameter variation: score a pre-loaded DataFrame (no disk read)
# `df` from Cell 2 is scored directly; passing it skips load_watchlist().
results_from_df = run_scan(min_score=3, df=df)
display(results_from_df)

# Same result as the disk-loaded path in Cell 3.
if results_from_df is None or results is None:
    raise ValueError('run_scan returned None — watchlist failed to load')
print('matches Cell 3:', results_from_df.equals(results))
```

```python
# Cell 5 — Parameter variation: cap the result size with top_n
top5 = run_scan(min_score=2, top_n=5)
display(top5)
print('rows returned:', 0 if top5 is None else len(top5))
```
